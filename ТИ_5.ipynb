```python
import numpy as np

# Параметры
n = 10                     # количество агентов
eps = 1e-6                 # точность сходимости
max_iter = 1000            # максимальное число итераций

# Генерация стохастической матрицы доверия
np.random.seed(42)         # для воспроизводимости
A = np.random.rand(n, n)
A = A / A.sum(axis=1, keepdims=True)

print("Матрица доверия A (стохастическая по строкам):")
print(np.round(A, 4))

# Моделирование без управления
x0_neutral = np.random.rand(n)          # начальные мнения
print("\nНачальные мнения (без управления):", np.round(x0_neutral, 3))

x = x0_neutral.copy()
for t in range(max_iter):
    x_new = A @ x
    if np.max(np.abs(x_new - x)) < eps:
        break
    x = x_new

print(f"Сходимость за {t+1} итераций.")
print("Итоговое мнение (без управления):", np.round(x, 4))

# Моделирование с информационным управлением
# Выбор непересекающихся множеств агентов для двух игроков
all_agents = set(range(n))
F = set(np.random.choice(n, size=2, replace=False))   # 2 агента для первого игрока
remaining = all_agents - F
S = set(np.random.choice(list(remaining), size=2, replace=False))  # 2 агента для второго
neutral = all_agents - F - S

# Управляющие воздействия
u = np.random.uniform(0, 100)      # для первого игрока (положительное)
v = np.random.uniform(-100, 0)     # для второго игрока (отрицательное)

# Начальные мнения с учётом управлений
x0_control = np.zeros(n)
for i in neutral:
    x0_control[i] = np.random.rand()          # нейтральные – случайные [0,1]
for i in F:
    x0_control[i] = u
for i in S:
    x0_control[i] = v

print("\n--- Информационное противоборство ---")
print(f"Агенты первого игрока (F): {sorted(F)} -> управление u = {u:.3f}")
print(f"Агенты второго игрока (S): {sorted(S)} -> управление v = {v:.3f}")
print(f"Нейтральные агенты: {sorted(neutral)}")
print("Начальные мнения с управлением:", np.round(x0_control, 3))

x = x0_control.copy()
for t in range(max_iter):
    x_new = A @ x
    if np.max(np.abs(x_new - x)) < eps:
        break
    x = x_new

print(f"Сходимость за {t+1} итераций.")
print("Итоговое мнение (с управлением):", np.round(x, 4))
```
