```python
from itertools import combinations

def shapley(N, v):
    n = len(N)
    phi = {i: 0 for i in N}
    for i in N:
        for k in range(n):
            for S in combinations([j for j in N if j != i], k):
                S = set(S)
                weight = (len(S) * (n - len(S) - 1)) / n
                # weight = |S|! (n-|S|-1)! / n!
                # for n=4: weight = |S|! (3-|S|)! / 24
                # используем факториалы вручную
                fact = {0:1,1:1,2:2,3:6}
                w = fact[len(S)] * fact[3 - len(S)] / 24
                phi[i] += w * (v[frozenset(S | {i})] - v[frozenset(S)])
    return phi

# Характеристическая функция v* как словарь
N = {1,2,3,4}
v = {
    frozenset(): 0,
    frozenset({1}): 3, frozenset({2}): 2, frozenset({3}): 3, frozenset({4}): 2,
    frozenset({1,2}): 6, frozenset({1,3}): 6, frozenset({1,4}): 5,
    frozenset({2,3}): 6, frozenset({2,4}): 4, frozenset({3,4}): 7,
    frozenset({1,2,3}): 10, frozenset({1,2,4}): 10,
    frozenset({1,3,4}): 10, frozenset({2,3,4}): 10,
    frozenset({1,2,3,4}): 13,
}
phi = shapley(N, v)
for i in sorted(phi):
    print(f"φ_{i} = {phi[i]:.4f} = {phi[i]}")
```
